```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = pd.read_csv('../datasource/fuelo_week_prices_all.csv')
data
```

```python
data = data.drop_duplicates()
```

```python
exchange_rate = 1.95583

#Извличаме годината от текста (последните 2 цифри)
data["year"] = data["Седмица"].str[-2:].astype(int) + 2000

#Избираме ценовите колони
price_cols = ["Бензин A95", "Дизел", "Пропан Бутан"]

#Конвертираме само до 2025
mask = data["year"] <= 2025
data.loc[mask, price_cols] = data.loc[mask, price_cols] / exchange_rate

# по желание закръгляне
data[price_cols] = data[price_cols].round(2)
```

```python
# data.to_csv("cleaned_fuelo_data.csv", index=False, encoding="utf-8-sig")
```

```python
data
```

```python
data["start_date"] = (
    data["Седмица"].str.split(" - ").str[0]
)

data["start_date"] = pd.to_datetime(
    data["start_date"],
    format="%d.%m"
)

data["start_date"] = data.apply(
    lambda row: row["start_date"].replace(year=row["year"]),
    axis=1
)

data
```
